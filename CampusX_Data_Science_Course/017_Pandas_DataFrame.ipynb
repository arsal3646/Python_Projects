```python
# how many matches did CSK team win in Kolkata

ipl_matches[(ipl_matches['City'] == 'Kolkata') & (ipl_matches['WinningTeam'] == 'Chennai Super Kings')]

# Note: this is not logical AND 'and' but we used & which is a bit-wise operator. 
# This is because we are working with Series objects and not single boolean values.

```

```python
ipl_matches[(ipl_matches['City'] == 'Kolkata') & (ipl_matches['WinningTeam'] == 'Chennai Super Kings')].shape[0]

# So there are 5 such matches where CSK won in Kolkata.
```

```python
# toss winner benefits the team or not? 
# we can check this by checking how many matches were won by the team that won the toss.

ipl_matches[ipl_matches['TossWinner'] == ipl_matches['WinningTeam']]
```

```python
ipl_matches.shape[0]

# Total number of matches played in IPL is 950. Out of these, 489 matches were won by the team that won the toss.
```

```python
ipl_matches[ipl_matches['TossWinner'] == ipl_matches['WinningTeam']].shape[0]

# In 489 matches out of 950 matches, the team that won the toss also won the match. This suggests that winning the toss may provide a significant advantage in IPL matches.
```

```python
ipl_matches[ipl_matches['TossWinner'] == ipl_matches['WinningTeam']].shape[0] / ipl_matches.shape[0]

# there is a very slight advantage of winning the toss in IPL matches 
# as the team that won the toss won the match in 51.47% of the cases. 
# However, this advantage is not very significant and may not be a decisive factor in determining the outcome of a match.
```

```python
# action movies with rating higher than 7.5
# we want to apply split

movies['genres']
```

```python
# first convert it to string. 





movies['genres'].str
```

```python
# then, split it using "|" 
 
movies['genres'].str.split("|")

# this will create a list such that 
```

```python
# then we can use apply() method to apply a function to each element of the Series.

mask1 = movies['genres'].str.split("|").apply(lambda x: 'Action' in x)
mask1

# what this does is that it will return a boolean Series where, 
# each element is True if the corresponding movie has 'Action' in its genres, and False otherwise.

# this is a very useful technique for filtering a DataFrame based on the values in a column that contains lists or strings.
```

```python
mask2 = movies['imdb_rating'] > 7.5
mask2
```

```python
movies[mask1 & mask2]

# why is the teacher using "mask" names? 
# it is a common convention to use the term "mask" when referring to a boolean Series that is used to filter a DataFrame.
```

```python
movies[mask1 & mask2].shape[0]
```

```python
# another way is to use the str.contains() method to check if the 'genres' column contains the string 'Action'.
# the code will be as below: 

movies[movies['genres'].str.contains('Action') & (movies['imdb_rating'] > 7.5)]
```
