```python
# Load the four datasets which we will be using in this notebook.

# value_counts is applicable both on Series and DataFrame. 
# It is used to count the number of unique values in a column.

marks = pd.DataFrame([
    [100,80,10],
    [90,70,7],
    [120,100,14],
    [80,70,14],
    [80,70,14]
    ], columns=['IQ','Marks','PackagePerYear']
    )

marks
```

```python
marks.value_counts()

# we can see that one row is repeated 2 times. So, the count of that row is 2.
```

```python
ipl.columns.tolist()
```

```python
# find which player won the most potm -- > in finals and qualifier matches. potm means player of the match.

# we need two columns, i.e. potm and match_type. 
# We will use these two columns to find the player who won the most potm in finals and qualifier matches.

ipl[ipl['MatchNumber'].str.isdigit()]['Player_of_Match'].value_counts()

# this is for non-finals and non-qualifier matches. We need to filter the data for finals and qualifier matches.
# just add NOT sign in the beginning of the condition.
```

```python
ipl[~ipl['MatchNumber'].str.isdigit()]['Player_of_Match'].value_counts()
```

```python
# graph (plot) for Toss decision, i.e. how many times teams decided to bat first and how many times they decided to field first.

ipl['TossDecision'].value_counts()
```

```python
ipl['TossDecision'].value_counts().plot(kind='pie')
```

```python
# HOW to ADD two series together if their index is same. 
# We can add two series together if their index is same. 
# If the index is not same, then it will return NaN for that index.

# using Team1 and Team2 columns, we will find the number of matches played by each team.

(ipl['Team1'].value_counts() + ipl['Team2'].value_counts()).sort_values(ascending=False)

# wow this is a very interesting method. It is used to find the number of matches played by each team.
```
