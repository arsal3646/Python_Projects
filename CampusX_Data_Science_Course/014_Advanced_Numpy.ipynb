# Advanced Numpy

```python
import numpy as np
import pandas as pd
# Why Numpy Array is preferred over Python List?
#   1. speed
#   2. memory consumption
#   3. convenience

# speed comparison between Python List and Numpy Array.

a = [i for i in range(10000000)]    # this is a Python List of 10 million integers starting from 0 to 9999999

b = [i for i in range(10000000, 20000000)]  # this is a Python List of 10 million integers starting from 10000000 to 19999999

# now we will add these two lists using Python List comprehension and measure the time taken for this operation.

c = []

# we need to record the time taken for this operation, so we will use the time module.
# so we will see the time before the loop starts and after the loop ends.

# current time is 'start'

import time

start = time.time()    # this will record the current time before the loop starts

for i in range(len(a)):      # len(a) and len(b) are both 10 million, so this loop will run 10 million times.
    c.append(a[i] + b[i])    # adding the corresponding elements of the two lists and appending to the new list c

time_taken_list = time.time() - start    # this will record the time after the loop ends and calculate the time taken for the operation

print(a[100])
print(b[100])
print(c[100])
print("Time taken by Python List:", time_taken_list)    # this will print the time taken for the operation in seconds
```

```python
# use Numpy Array to perform the same operation and measure the time taken for this operation.

a = np.arange(10000000)    # this is a Numpy Array of 10 million integers starting from 0 to 9999999
b = np.arange(10000000, 20000000)  # this is a Numpy Array of 10 million integers starting from 10000000 to 19999999

start_Numpy = time.time()    # this will record the current time before the operation

c = a + b    # adding the two Numpy Arrays

time_taken_Numpy = time.time() - start_Numpy    # this will record the time after the operation and calculate the time taken for the operation

print(a[100])
print(b[100])
print(c[100])
print("Time taken by Numpy Array:", time_taken_Numpy)    # this will print the time taken for the operation in seconds

time_taken_Numpy
```

```python
# Compare time taken by Numpy vs Python List for the same operation.

time_taken_list/time_taken_Numpy

# Two reasons why Numpy Array is faster than Python List:

# 1. Numpy Arrays are implemented in C and optimized for performance, 
#   while Python Lists are implemented in Python and have more overhead.  

# 2. Numpy Arrays are homogeneous, meaning they can only contain elements of the same data type, 
#    which allows for more efficient memory usage and faster computations. 
#    Python Lists can contain elements of different data types, which adds overhead and slows down operations.    
```

```python
# Memory

# Now let us see how much memoryview is consumed by Python List and Numpy Array for the same data.

a = [i for i in range(10000000)]    # this is a Python List of 10 million integers starting from 0 to 9999999
import sys
size_a = sys.getsizeof(a)    # this will return the size of the Python List in bytes
size_a
```

```python
# let us create a Numpy Array of 10 million integers starting from 0 to 9999999 with data type int32 instead of int64 (default data type for Numpy Array).
revised_a = np.arange(10000000, dtype=np.int32)
size_revised_a = revised_a.nbytes
size_revised_a
```

```python
from numpy import half


size_a / size_revised_a

# we reduced the size to half by changing the data type of the Numpy Array to int32 instead of int64.
```

```python
# just let us see how int8 can make problem for us.

a_int8 = np.arange(1000, dtype=np.int8)
a_int8

# As you can see, it reached 127 and then went straight to -128 and then started increasing again. 
# This is because the range of int8 is -128 to 127, and when it exceeds 127, it wraps around to -128. 
# This is called overflow.

# Same is in int16 also. As soon as it reaches 32767, it will wrap around to -32768. 
```

```python
# Boolean Indexing

# we dont need always to use an index to fetch the data from an array. 
# We can also use a "condition" or "logic" to fetch the data from an array. This is called Boolean Indexing.

d = np.random.randint(1,100,24).reshape(6,4)   # this will create a 2D Numpy Array of shape (6, 4) with random integers between 1 and 100 
d
```

```python
d > 70   
# this will return a boolean array of the same shape as array_bool_ind_a, 
# where each element is True if the corresponding element in array_bool_ind_a is greater than 50, and False otherwise.
```

```python
# we can use this boolean array as a "mask" to fetch the elements from the original array that satisfy the condition.

# Only true values will be fetched from the original array, and false values will be ignored.

d[d > 70]
```

```python
# Boolean indexing (e.g. Fetch odd/even numbers)

d[d % 2 == 0]   
# this will return all the even numbers from the array_bool_ind_a 
```

```python
# boolean indexing (e.g. Fetch even numbers which are greater than 50)
d[(d % 2 == 0) & (d > 50)]
```

```python
# Why we used "&" instead of "and" in the above example?
# because "and" is a logical operator that works on boolean values, while "&" is a bitwise operator that works on binary values.
# in the above example, we are using "&" to combine two boolean arrays, which is a bitwise operation.

# to remember:
# "and" is a logical operator that works on boolean values, while "&" is a bitwise operator that works on binary values. 
# in the above example, we are using "&" to combine two boolean arrays, which is a bitwise operation.

```

```python
# find numbers divisible by 7

d[(d % 7 == 0)]
```

```python
# find numbers not divisible by 7
d[(d%7 != 0)]
```

```python
# another way it to put "not" operator in front of the condition, i.e. 
d[~(d % 7 == 0)]
```

# Broadcasting

A very important but confusing topic.

```python
# Broadcasting

# This technique allows you to perform operations on arrays of different shapes and sizes, 
# without the need for explicit looping or reshaping.
```

```python
# The term "broadcasting" refers to the ability of Numpy to perform arithmetic operations on arrays of different shapes and sizes,
# by "stretching" the smaller array to match the shape of the larger array,
# without actually copying the data in memory. 

# This is done by "broadcasting" the smaller array across the larger array, 
# so that they have compatible shapes for the operation.
```

```python
# same shape

e = np.arange(6).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 0 to 5

f = np.arange(6,12).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 6 to 11
```

```python
e
```

```python
f
```

```python
# e and f have the same shape, so we can perform arithmetic operations on them without any issues.
```

```python
e+f
```

```python
e-f
```

```python
e*f
```

```python
e/f
```

```python
e%f
```

```python
# different shape

g = np.arange(6).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 0 to 5

h = np.arange(3).reshape(1, 3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 6 to 11
```

```python
g
```

```python
h
```

```python
# can we add these two arrays of different shapes? 
# Yes, we can add these two arrays of different shapes using broadcasting.

# in normal mathematics, we cannot add two matrices of different shapes.

g + h  

# this will work in Numpy due to broadcasting
# what Numpy will do is that it will "stretch" the smaller array h to match the shape of the larger array g,
# by repeating the values of h along the first axis, so that they have compatible shapes for the operation.


```
