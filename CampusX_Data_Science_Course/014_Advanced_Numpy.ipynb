```python
import numpy as np
import pandas as pd
# Why Numpy Array is preferred over Python List?
#   1. speed
#   2. memory consumption
#   3. convenience

# speed comparison between Python List and Numpy Array.

a = [i for i in range(10000000)]    # this is a Python List of 10 million integers starting from 0 to 9999999

b = [i for i in range(10000000, 20000000)]  # this is a Python List of 10 million integers starting from 10000000 to 19999999

# now we will add these two lists using Python List comprehension and measure the time taken for this operation.

c = []

# we need to record the time taken for this operation, so we will use the time module.
# so we will see the time before the loop starts and after the loop ends.

# current time is 'start'

import time

start = time.time()    # this will record the current time before the loop starts

for i in range(len(a)):      # len(a) and len(b) are both 10 million, so this loop will run 10 million times.
    c.append(a[i] + b[i])    # adding the corresponding elements of the two lists and appending to the new list c

time_taken_list = time.time() - start    # this will record the time after the loop ends and calculate the time taken for the operation

print(a[100])
print(b[100])
print(c[100])
print("Time taken by Python List:", time_taken_list)    # this will print the time taken for the operation in seconds
```

```python
# use Numpy Array to perform the same operation and measure the time taken for this operation.

a = np.arange(10000000)    # this is a Numpy Array of 10 million integers starting from 0 to 9999999
b = np.arange(10000000, 20000000)  # this is a Numpy Array of 10 million integers starting from 10000000 to 19999999

start_Numpy = time.time()    # this will record the current time before the operation

c = a + b    # adding the two Numpy Arrays

time_taken_Numpy = time.time() - start_Numpy    # this will record the time after the operation and calculate the time taken for the operation

print(a[100])
print(b[100])
print(c[100])
print("Time taken by Numpy Array:", time_taken_Numpy)    # this will print the time taken for the operation in seconds

time_taken_Numpy
```

```python
# Compare time taken by Numpy vs Python List for the same operation.

time_taken_list/time_taken_Numpy

# Two reasons why Numpy Array is faster than Python List:

# 1. Numpy Arrays are implemented in C and optimized for performance, 
#   while Python Lists are implemented in Python and have more overhead.  

# 2. Numpy Arrays are homogeneous, meaning they can only contain elements of the same data type, 
#    which allows for more efficient memory usage and faster computations. 
#    Python Lists can contain elements of different data types, which adds overhead and slows down operations.    
```

```python
# Boolean Indexing

# we dont need always to use an index to fetch the data from an array. 
# We can also use a "condition" or "logic" to fetch the data from an array. This is called Boolean Indexing.

d = np.random.randint(1,100,24).reshape(6,4)   # this will create a 2D Numpy Array of shape (6, 4) with random integers between 1 and 100 
d
```

# Broadcasting

A very important but confusing topic.

```python
# Broadcasting

# This technique allows you to perform operations on arrays of different shapes and sizes, 
# without the need for explicit looping or reshaping.
```

```python
# The term "broadcasting" refers to the ability of Numpy to perform arithmetic operations on arrays of different shapes and sizes,
# by "stretching" the smaller array to match the shape of the larger array,
# without actually copying the data in memory. 

# This is done by "broadcasting" the smaller array across the larger array, 
# so that they have compatible shapes for the operation.
```

```python
# same shape

e = np.arange(6).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 0 to 5

f = np.arange(6,12).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 6 to 11
```

```python
e
```

```python
f
```

```python
# e and f have the same shape, so we can perform arithmetic operations on them without any issues.
```

```python
e+f
```

```python
e-f
```

```python
e*f
```

```python
e/f
```

```python
e%f
```

```python
# different shape

g = np.arange(6).reshape(2, 3)    # this will create a 2D Numpy Array of shape (2, 3) with values from 0 to 5

h = np.arange(3).reshape(1, 3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 6 to 11
```

```python
g
```

```python
h
```

```python
# can we add these two arrays of different shapes? 
# Yes, we can add these two arrays of different shapes using broadcasting.

# in normal mathematics, we cannot add two matrices of different shapes.

g + h  

# this will work in Numpy due to broadcasting
# what Numpy will do is that it will "stretch" the smaller array h to match the shape of the larger array g,
# by repeating the values of h along the first axis, so that they have compatible shapes for the operation.
```

```python
# In the above example, the row 0,1,2 was replicated in second row to match the shape of the larger array g, 
# so that they have compatible shapes for the operation.
```

```python
# Three rules of broadcasting:

# 1. Make the two arrays have the same number of dimensions by adding new dimensions of size 1 to the smaller array, if necessary.
 
# 2. If the shape of the two arrays does not match in any dimension, 
#    the array with shape equal to 1 in that dimension is stretched to match the other shape.

# 3. If in any dimension the sizes disagree and neither is equal to 1, an error is raised.

```

```python
# Understand Rule 1 of broadcasting:

# For example, you have 2D array of shape (3,2) and 1D array of shape (3,). 
# According to Rule 1, you need to reshape the 1D array to (1,3) before broadcasting.
# So the 1D array will become a 2D array of shape (1,3) by repeating the values along the first axis.

# For example, you have a 3D array of shape (3,3,3) and 1D array of shape (3,).
# According to Rule 1, 
#      you need to reshape the 1D array to (1,1,3) before broadcasting, 
#      i.e. 1 is added to the "head" of the shape of the 1D array to make it compatible with the 3D array.
#      So the 1D array will become a 3D array of shape (1,1,3) by repeating the values along the first and second axes.
```

```python
# Rule 2 of broadcasting:

# According to Rule 2, if the shape of the two arrays does not match in any dimension,
# the array with shape equal to 1 in that dimension is stretched to match the other shape.

# Rule 3 of broadcasting: 
# Error is raised (broadcasting fails) if in any dimension the sizes disagree and neither is equal to 1.
```

```python
# Example: Broadcasting

i = np.arange(12).reshape(4,3)    # this will create a 2D Numpy Array of shape (4, 3) with values from 0 to 11

j = np.arange(3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 0 to 2


print("Array i is:\n", i)
print("\n\nArray j is:\n", j)
```

```python
sum_i_j = i+j

sum_i_j

# what happened actually?

# j became:
# [[0, 1, 2],
#  [0, 1, 2],
#  [0, 1, 2],
#  [0, 1, 2]]

# then it was added to i
```

```python
# So, i array is of shape (4,3) and j array is of shape (1,3).

# Now, you can broadcast j array to i array.

# how this will work?
# Step 1: add a new dimension of size 1 to the smaller array j, so that it becomes a 2D array of shape (1,3).

# Step 2: Broadcast the 1 to 4, and we get new size of j array as (4,3).

# Check: are the shapes same? 
# if yes, then we can perform arithmetic operations on these two arrays of different shapes using broadcasting.

# Yes, because we get new size of j array as (4,3) after broadcasting, which is same as the size of i array.

# Now, you will be able to do arithmetic operations on these two arrays of different shapes using broadcasting.
```

```python
# Modifying above example: Broadcasting

k = np.arange(12).reshape(3,4)    # this will create a 2D Numpy Array of shape (3, 4) with values from 0 to 11

l = np.arange(3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 0 to 2

print("Array k is:\n", k)
print("\n\nArray l is:\n", l)
```

```python
# k+l will give error
# broadcasting will not work in this case because the shapes of the two arrays are not compatible for broadcasting.

# what happened actually? 

# j became:
# [[0, 1, 2],
#  [0, 1, 2],
#  [0, 1, 2]]

# The steps followed by j are:
# Step 1: add a new dimension of size 1 to the smaller array l, so that it becomes a 2D array of shape (1,3).
# Step 2: Broadcast the 1 to 3, and we get new size of l array as (3,3).
# Check: are the shapes same?
# if yes, then we can perform arithmetic operations on these two arrays of different shapes using broadcasting 
# if no, then we cannot perform arithmetic operations on these two arrays of different shapes using broadcasting.
```

```python
# Third example: Broadcasting

# Modifying above example: Broadcasting

m = np.arange(3).reshape(1,3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 0 to 2

n = np.arange(3).reshape(3,1)    # this will create a 2D Numpy Array of shape (1, 3) with values from 3 to 5

print("Array m is:\n", m)
print("\n\nArray n is:\n", n)
```

```python
# So, we have (1,3) and (3,1) of 2D shapes.
# We can break this down into 3 steps to see how broadcasting will work in this case.
# Step 1: add a new dimension of size 1 to the smaller array m, so that it becomes a 2D array of shape (1,3).
# Step 2: Broadcast the 1 to 3, and we get new size of m array as (3,3).
# Step 3: Broadcast the 1 to 3, and we get new size of n array as (3,3).
```

```python
m+n
```

```python
# what happened actually?
# m became:
# [[0, 1, 2],
#  [0, 1, 2],
#  [0, 1, 2]]   

# n:
#[[0, 0, 0],
#  [1, 1, 1],
#  [2, 2, 2]]
```

```python
# Fourth example: Broadcasting

# Modifying above example: Broadcasting

p = np.arange(3).reshape(1,3)    # this will create a 2D Numpy Array of shape (1, 3) with values from 0 to 2

q = np.arange(4).reshape(4,1)    # this will create a 2D Numpy Array of shape (4, 1) with values from 0 to 3

print("Array p is:\n", p)
print("\n\nArray q is:\n", q)
```

```python
# both p and q are 2D arrays, but they have different shapes.

# p has shape (1, 3) and q has shape (4, 1)

# p is changed to (4,3) and q is changed to (4,3) after broadcasting. 
# That is why we can perform arithmetic operations on these two arrays of different shapes using broadcasting.
```

```python
p + q
```

```python
# mathematical functions in Numpy

a = np.arange(10)    # this will create a 1D Numpy Array of shape (10,) with values from 0 to 9
np.sin(a)
```

```python
# We can do almost every mathematical operation on Numpy Array, like sin, cos, tan, exp, log, sqrt, etc.
# let us see sigmoid as an example of the power/strength of Numpy Array.

# sigmoid function is a very useful function in machine learning and deep learning.
# It is a mathematical function that takes a real-valued number and maps it to a value between 0 and 1.

# Formula for sigmoid function is:
# sigma(x) = 1 / (1 + e^{-x})

def sigmoid(array):
    return 1 / (1 + np.exp(-array))

print(sigmoid(a))

# sigmoid is from 0 to 1, and it is a smooth function.
# smooth function means that it is differentiable, and its derivative is continuous.

# we can draw the sigmoid function using matplotlib.
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 100)  # this will create a 1D Numpy Array of shape (100,) with values from -10 to 10
y = sigmoid(x)
plt.plot(x, y)
plt.title("Sigmoid Function")
plt.xlabel("x")
plt.ylabel("sigma(x)")
plt.show()

```

```python
# mean squared error (MSE) is a very useful function in machine learning and deep learning.
# It is a mathematical function that takes two real-valued numbers and maps it to a value between 0 and infinity.

# Formula for mean squared error is:
# MSE = (1/n) * Σ(y_true - y_pred)^2

# Loss function is a mathematical function that takes two real-valued numbers and maps it to a value between 0 and infinity.
# this is same as mean squared error, but it is used in machine learning and deep learning to measure the performance of a model.

# Loss function is like a candle/lamp in the dark, it shows us the way to reach our destination.
# it will tell how can we reduce the distance (error) between the predicted value and the actual value 
# so that we can reach our destination (minimum error).

# the first loss function is the mean squared error (MSE) loss function, which is used in linear regression problems.

# let us create two arrays as below:

actual = np.random.randint(1, 50, 25)  # this will create a 1D Numpy Array of shape (25,) with random integers between 1 and 50

predicted = np.random.randint(1, 50, 25)  # this will create a 1D Numpy Array of shape (25,) with random integers between 1 and 50


print(actual)    # actual means actual values of the target variable in the dataset, i.e. true values of the target variable.
print(predicted) # predicted means predicted by the model   
```

```python
# what we need to do is to caculate e.g. 1st element of MSE = (actual[0] - predicted[0])^2
# then we will sum all the elements of MSE and divide it by the number of elements in the array to get the mean squared error.

# e.g. if first element of actual is 10 and first element of predicted is 12, 
# then the first element of MSE will be (10-12)^2 = 4

def mean_squared_error(x, y):
    return np.mean((x - y) ** 2)   # the ** means power or exponentiation
```

```python
# call the function to calculate the mean squared error between the actual and predicted values.
print(mean_squared_error(actual, predicted))
```

```python
# Now comes the big headache, i.e. "Missing Values" or "NaN" values in the dataset.

a1 = np.array([1, 2, 3, np.nan, 5, 6, np.nan, 8, 9, 10])   # this will create a 1D Numpy Array of shape (10,) with values from 1 to 10 and two NaN values
a1
```

```python
# nan and none are two different things in Python.

# none is not a missing value, it is a special value that represents the absence of a value or a null value.
# np.nan is a special value that represents a missing value or a not a number value.

# shortcut way done by beginner data science people is that they just remove 
# the values of the rows or columns which have NaN values in them, 
# but this is not a good practice because it will lead to loss of data and information.

np.isnan(a1)   # this will return a boolean array of the same shape as a1, where each element is True if the corresponding element in a1 is NaN, and False otherwise.
```

```python
a1[np.isnan(a1)]   
    
# this will return the elements of a1 that are NaN.
```

```python
# but we want the ones which are not NaN, so we can use the "not" operator "~" in front of the condition.

a1[~np.isnan(a1)]   # this will return the elements of a1 that are not NaN.
```
