```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Recap

# Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. 
# A Series is like a fixed-size dict in that you can get and set values by index label.  

# DataFrame is a two-dimensional labeled data structure with columns of potentially different types. 
# You can think of it like a spreadsheet or SQL table, or a dict of Series objects. 
# It is generally the most commonly used pandas object.  


# for fetching any info from Series, we just need SINGLE index, 
# but for DataFrame, we need TWO indexes (row and column) to fetch any info.
```

```python
# Can we have multiple indexes for Series and DataFrame? 
# Yes, we can have multiple indexes for both Series and DataFrame.

# NOT Recommended way is as below:

index_val = [('cse',2019),('cse',2020),('cse',2021),('cse',2022),('ece',2019),('ece',2020),('ece',2021),('ece',2022)]
```

```python
# now convert the above to a Pandas Series

# NOTE: The index is a combination of two values, so we can say that it is a MultiIndex Series.

a = pd.Series([1,2,3,4,5,6,7,8], index=index_val)
a
```

```python
# let us try to use the above series

a[('cse',2020)]
```

```python
a[('ece',2020)]
```

```python
# a[('ece)] will give error because ece or cse is not an index, it is a part of index.
# If someone asks use to find something for all CSE we will be in trouble because we cannot use a single index to fetch all CSE values.

# because of this issue this methodology is not recommended, instead we should use MultiIndex Series and DataFrame.
```
