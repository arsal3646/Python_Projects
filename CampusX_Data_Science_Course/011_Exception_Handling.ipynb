```python
# Errors are of two types: Syntax errors and Exceptions.

# Syntax errors are errors in the structure of the code, which prevent the code from being executed.
#  For example, missing a colon at the end of a function definition or using incorrect indentation.

# Exceptions are errors that occur during the execution of the code, even if the code is syntactically correct.
#  For example, dividing by zero or trying to access a non-existent file.

# Syntax errors are detected by the Python interpreter before the code is executed
# Exceptions are detected during the execution of the code.
```

```python
# Some examples of syntax errors:

# print "Hello, World!"  # Missing parentheses
# if True  # Missing colon

# iff True:  # Misspelled 'if'
# for i in range(5)  # Missing colon

# def my_function()  # Missing colon
# import non_existent_module  # Trying to import a non-existent module
# import mathh # Misspelled 'math' module

# Key Error: Trying to access a non-existent key in a dictionary, e.g., my_dict['non_existent_key']

# Type Error: Performing an operation on incompatible data types, e.g., adding a string and an integer.

# Value Error: Passing an argument of the wrong type to a function, 
# e.g., int('abc') - it was expected to be a string that can be converted to an integer, but 'abc' cannot be converted.

# Attribute Error: Trying to access an attribute or method that does not exist for a particular object,
# e.g., my_list.non_existent_method() - 'list' object has no attribute 'non_existent_method'

# Name Error: Trying to use a variable or function that has not been defined, 
# e.g., print(undefined_variable) - 'undefined_variable' is not defined.

# Index Error: Trying to access an index that is out of range for a list or string, 
# e.g. my_list[10] - list index out of range.
```
