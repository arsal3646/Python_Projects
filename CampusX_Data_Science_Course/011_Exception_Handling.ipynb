```python
# Errors are of two types: Syntax errors and Exceptions.

# Syntax errors are errors in the structure of the code, which prevent the code from being executed.
#  For example, missing a colon at the end of a function definition or using incorrect indentation.

# Exceptions are errors that occur during the execution of the code, even if the code is syntactically correct.
#  For example, dividing by zero or trying to access a non-existent file.

# Syntax errors are detected by the Python interpreter before the code is executed
# Exceptions are detected during the execution of the code.
```

```python
# Some examples of syntax errors:

# print "Hello, World!"  # Missing parentheses
# if True  # Missing colon

# iff True:  # Misspelled 'if'
# for i in range(5)  # Missing colon

# def my_function()  # Missing colon
# import non_existent_module  # Trying to import a non-existent module
# import mathh # Misspelled 'math' module

# Key Error: Trying to access a non-existent key in a dictionary, e.g., my_dict['non_existent_key']

# Type Error: Performing an operation on incompatible data types, e.g., adding a string and an integer.

# Value Error: Passing an argument of the wrong type to a function, 
# e.g., int('abc') - it was expected to be a string that can be converted to an integer, but 'abc' cannot be converted.

# Attribute Error: Trying to access an attribute or method that does not exist for a particular object,
# e.g., my_list.non_existent_method() - 'list' object has no attribute 'non_existent_method'

# Name Error: Trying to use a variable or function that has not been defined, 
# e.g., print(undefined_variable) - 'undefined_variable' is not defined.

# Index Error: Trying to access an index that is out of range for a list or string, 
# e.g. my_list[10] - list index out of range.
```

```python
# Exceptions cant be detected by the Python interpreter before the code is executed, 
# i.e., they are detected during the execution of the code.

# Some examples of exceptions (runtime errors):

# print(1 / 0)  # Division by zero

# open("non_existent_file.txt")  # File not found   

# memory overflow error: Trying to allocate more memory than is available, e.g., creating a very large list or array.
# This happens a lot in deep learning when the model is too large for the available GPU memory.

# Database connection error: Trying to connect to a database that is not available or has incorrect credentials.

# Stack overflow error: 
# This occurs when there is too much recursion in the code, leading to the call stack exceeding its limit.
# This is a way that Python is communicating with the programmer that the code is not efficient and needs to be optimized.
# However, it not good that end-users see this error, so it is important to handle it properly in production code.
# Also, for security reasons, it is important to handle this error properly, 
# as it can be exploited by attackers to crash the application.

# Let us understand Stack overflow error with an example:
# def recursive_function():
#     recursive_function()  # This will eventually lead to a stack overflow error

def recursive_function():
    recursive_function()  # This will eventually lead to a stack overflow error
```

```python
# Now let us see how to handle exceptions in Python. 
# In java we used to call it try-catch block, in Python we call it try-except block.

# we need to use try-except blocks to handle exceptions in Python. 
# The code that may raise an exception is placed inside the try block, 
# and the code to handle the exception is placed inside the except block.

with open("sample3.txt",'w') as f:
    f.write("Hello, World!")  

# This will not raise an exception because the file is opened in write mode.
# if 'w' was not written in the open function, it would have raised an exception because the file does not exist.
# also, default mode is 'r' which is read mode, so if the file does not exist, it will raise an exception.
```
