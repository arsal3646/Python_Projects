```python
# I/O stands for Input/Output.
# Input means taking data from the user
# Output means displaying data to the user.

# In Python everything is Unicode, so we can use any language to take input and display output.
```

```python
# There are two types of files in Python:
# 1. Text files
# 2. Binary files
```

```python
# For example, a text file can be a .txt file, a .csv file, or a .json file.

# In a text file, the data is stored in a human-readable format, and it can be opened and edited using a text editor.
# All text files are binary files, but not all binary files are text files.
# All text files are binary files because they are stored in a binary format, 
# but not all binary files are text files because they may contain data that is not human-readable.

# All text files are in UTF-8 encoding, which is a standard for encoding text in computers.
# While binary files can be in any encoding, such as UTF-16, ASCII, or any other encoding.

# All program files are text files, but not all text files are program files.
```

```python
# For example, a binary file can be a .jpg or .png file.

# These are binary files because they contain data that is not human-readable, 
# and they can be opened and edited using a binary editor.

# Many examples of binary files are 
#   1. images
#   2. videos
#   3. audio files
#   4. executable files
```

```python
# Difference between text editor and binary editor:
# A text editor is a program that allows you to create and edit text files.
# A binary editor is a program that allows you to create and edit binary files.
```

```python
# Difference between text file and binary file:
# A text file is a file that contains data that is human-readable,
# while a binary file is a file that contains data that is not human-readable.  
```

```python
# Python provides built-in functions to work with files. 
# The most commonly used functions are 
#   1. open()
#   2. read()
#   3. write()
#   4. close()
```
