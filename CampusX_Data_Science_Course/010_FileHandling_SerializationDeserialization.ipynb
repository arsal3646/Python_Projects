```python
# use context manager (WITH)

# "with" is a context manager that automatically closes the file after the block of code is executed 
# even if an exception is raised.

# what is context manager?
# A context manager is a Python object that defines the runtime context to be established when executing a block of code. 

# It is used to manage resources such as 
#   1. file streams
#   2. network connections
#   3. database connections 

# The context manager is responsible for 
#   1. setting up the context, ensuring that resources are properly acquired.
#   2. tearing down the context, ensuring that resources are properly released.
```

```python
# moving within a file --> 10 char then go back to 0 char then read 10 char again. 
# Both times it will read the same 10 characters.

with open('sample2.txt', 'r') as f:
    print(f.read(10))    # read first 10 characters
    f.seek(0)            # move the cursor to the beginning of the file
    print(f.read(10))    # read next 10 characters
   
```

```python
# moving within a file --> 10 char then 10 char

with open('sample2.txt', 'r') as f:
    print(f.read(10))    # read first 10 characters
    print(f.read(10))    # read next 10 characters

# how is this working? This is because of role of BUFFER. 
# The buffer is a temporary storage area in memory that holds data that is being transferred between two locations.
```

```python
# the above BUFFER technique is used to 
#   1. improve the performance of file I/O operations by reducing the number of disk accesses required to read or write data.
#   2. provide a way to read or write data in chunks, which can be useful when working with large files or streams of data.

# Benefit # 2 allows you to upload big files even if RAM size is small. 
# Because the file is read in chunks, and each chunk is processed before the next chunk is read.

# see below demonstration of the above point.
```

```python
# let us create a big file in memory

big_L = ['Hello World!' for i in range(1000)]   # create a list strings printed many times to make it big.

with open('big_file.txt', 'w') as f:    # open the file in write mode
    f.writelines(big_L)  # write the list of strings to the file
```

```python
with open('big_file.txt', 'r') as f:    # open the file in read mode
    chunk_size = 100  # define the chunk size

    while len(f.read(chunk_size))>0: # check if the chunk is not empty which means we continue to process the file
        print(f.read(chunk_size))
        f.read(chunk_size)  # read the next chunk of data from the file
```

```python
# after every ten char we place dummy values ****** just to see the effect of BUFFER.

with open('big_file.txt', 'r') as f:    # open the file in read mode
    chunk_size = 50  # define the chunk size

    while len(f.read(chunk_size))>0: # check if the chunk is not empty which means we continue to process the file
        print(f.read(chunk_size), end='*****')  # print the chunk and end it with *****
        f.read(chunk_size)  # read the next chunk of data from the file
```

```python
with open('big_file.txt', 'r') as f:  # open the file in read mode
    print(f.seek(0))     # move the cursor to the beginning of the file
    print(f.seek(100))   # move the cursor to the 100th character of the file
```

```python
with open('big_file.txt', 'r') as f:  # open the file in read mode
    print(f.tell())     # get the current position of the cursor in the file
    f.seek(65)          # move the cursor to the 65th character of the file
    print(f.tell())     # get the current position of the cursor in the file after moving it to the 65th character
```
